```python
def sip_future_value(P, R, n, annuity_due=False):
    """Calculate the future value of a Systematic Investment Plan (SIP).
    
    Arguments:
        P (float): SIP amount invested each month
        R (float): Expected annual return in % form (12%)
        n (int): Investment duration in years
        annuity_due (bool): True if SIP is invested at the start of each month, False for end of month
    
    Returns:
        dict: Dictionary containing FV, invested, and gain
    """
    R = R / 100  # convert percentage to decimal
    m = 12 * n  # total number of months
    i = (1 + R)**(1/12) - 1  # effective monthly return
    
    # Future Value calculation
    fv = P * ((1 + i)**m - 1) / i
    if annuity_due:
        fv *= (1 + i)
    
    invested = P * m
    gain = fv - invested
    
    return {
        'future_value': float(f"{fv:.4f}"),
        'invested': float(f"{invested:.4f}"),
        'gain': float(f"{gain:.4f}")
    }
```

# Step-Up SIP

```python
def sip_future_value_step_up(P, R, n, step_up=10, annuity_due=False):
    """Future value of SIP with annual step-up in monthly SIP amount."""
    R = R / 100
    step = step_up / 100
    i = (1 + R) ** (1 / 12) - 1
    total_months = n * 12

    fv = 0.0
    current_monthly = P
    for month in range(total_months):
        if month > 0 and month % 12 == 0:
            current_monthly *= (1 + step)

        if annuity_due:
            # contribution at start of month
            fv = (fv + current_monthly) * (1 + i)
        else:
            # contribution at end of month
            fv = fv * (1 + i) + current_monthly

    invested = 0.0
    current_yearly = P
    for _ in range(n):
        invested += current_yearly * 12
        current_yearly *= (1 + step)

    gain = fv - invested

    return {
        "future_value": float(f"{fv:.4f}"),
        "invested": float(f"{invested:.4f}"),
        "gain": float(f"{gain:.4f}"),
    }
```

# SWP

```python
def swp_future_value(P=10000, R=12, n=10, monthly_withdrawal=1000, annuity_due=False):
    """Systematic Withdrawal Plan (SWP) from a corpus.
    
    P : initial corpus (principal), default 10000
    R : annual return in %, e.g. 12
    n : duration in years
    monthly_withdrawal : amount withdrawn each month
    annuity_due : if True, withdrawal at start of month (then growth), else end of month
    """
    R = R / 100
    i = (1 + R) ** (1 / 12) - 1
    months = n * 12
    
    corpus = float(P)
    total_withdrawn = 0.0
    actual_months = 0
    
    for month in range(months):
        actual_months += 1
        if annuity_due:
            # withdraw first, then grow
            corpus -= monthly_withdrawal
            total_withdrawn += monthly_withdrawal
            if corpus <= 0:
                corpus = 0.0
                break
            corpus *= (1 + i)
        else:
            # grow first, then withdraw
            corpus *= (1 + i)
            corpus -= monthly_withdrawal
            total_withdrawn += monthly_withdrawal
            if corpus <= 0:
                corpus = 0.0
                break
    
    result = {
        "remaining_corpus": float(f"{corpus:.4f}"),
        "total_withdrawn": float(f"{total_withdrawn:.4f}"),
        "months_used": actual_months
    }
    
    if actual_months < months:
        result['warning'] = f"Corpus emptied after {actual_months} months. SWP not possible for full duration."
    
    return result
```

# SIP Gaps/Pauses

```python
def sip_future_value_with_gaps(P, R, n, GY, position, annuity_due=False):
    """Calculate the future value of a Systematic Investment Plan (SIP) with gap years.
    
    Arguments:
        P (float): SIP amount invested each month
        R (float): Expected annual return in % form (12%)
        n (int): Total investment duration in years
        GY (int): Number of gap years (years with no investment)
        position (int): Starting year of the gap (1-based index, e.g., 3 means gap starts at year 3)
        annuity_due (bool): True if SIP is invested at the start of each month, False for end of month
    
    Returns:
        dict: Dictionary containing FV, invested, and gain
    """
    R = R / 100  # convert percentage to decimal
    m = 12 * n  # total number of months
    i = (1 + R)**(1/12) - 1  # effective monthly return
    
    # Calculate months with gaps
    gap_start_month = (position - 1) * 12
    gap_end_month = gap_start_month + GY * 12
    
    fv = 0.0
    invested = 0.0
    
    for month in range(m):
        if month < gap_start_month or month >= gap_end_month:
            invested += P
            if annuity_due:
                fv = (fv + P) * (1 + i)
            else:
                fv = fv * (1 + i) + P
        else:
            fv *= (1 + i)  # grow without investment
    
    gain = fv - invested
    
    return {
        'future_value': float(f"{fv:.4f}"),
        'invested': float(f"{invested:.4f}"),
        'gain': float(f"{gain:.4f}")
    }
```

# SWP Gaps/Pauses

```python
def swp_future_value_with_gaps(P, R, n, W, GY, position, annuity_due=False):
    """Calculate the remaining corpus of a Systematic Withdrawal Plan (SWP) with gap years.
    
    Arguments:
        P (float): Initial corpus (principal)
        R (float): Expected annual return in % form (12%)
        n (int): Total withdrawal duration in years
        W (float): Monthly withdrawal amount
        GY (int): Number of gap years (years with no withdrawal)
        position (int): Starting year of the gap (1-based index)
        annuity_due (bool): True if withdrawal at start of month, False for end of month
    
    Returns:
        dict: Dictionary containing remaining_corpus, total_withdrawn, and months_used
    """
    R = R / 100
    i = (1 + R) ** (1 / 12) - 1
    months = n * 12
    
    gap_start_month = (position - 1) * 12
    gap_end_month = gap_start_month + GY * 12
    
    corpus = float(P)
    total_withdrawn = 0.0
    actual_months = 0
    
    for month in range(months):
        actual_months += 1
        
        if month < gap_start_month or month >= gap_end_month:
            # withdrawal period
            if annuity_due:
                corpus -= W
                total_withdrawn += W
                if corpus <= 0:
                    corpus = 0.0
                    break
                corpus *= (1 + i)
            else:
                corpus *= (1 + i)
                corpus -= W
                total_withdrawn += W
                if corpus <= 0:
                    corpus = 0.0
                    break
        else:
            # gap period - only growth
            corpus *= (1 + i)
    
    return {
        'remaining_corpus': float(f"{corpus:.4f}"),
        'total_withdrawn': float(f"{total_withdrawn:.4f}"),
        'months_used': actual_months
    }
```
